```python
import ollama
```

```python
model_name = "llama3.1:latest"

system_msg = '''
Tarefa: Você é um profissional de um hospital especializado na análise de prontuários médicos, com profunda experiência na identificação de materiais utilizados em procedimentos cirúrgicos.

Instrução: A partir do prontuário médico fornecido a seguir, extraia os materiais que foram utilizados no procedimento cirúrgico com clareza e precisão. 
'''
        
final_remarks = '''
      O seu retorno deve ser uma lista de materiais utilizados no procedimento cirúrgico. Cada item deve seguir a estrutura a seguir:

{
    "item": "resumo sucinto do que se trata o material",
    "procedimento": "breve descrição do procedimento realizado com o material"
} '''

#final_remarks = "O seu retorno deve ser uma lista de materiais utilizados no procedimento cirúrgico"

for i in range(1, 8):
    # lendo o arquivo txt
    txt = open(f'medical-abstracts/medical_abstract_{i}.txt', 'r').read()

    prompt = f"{system_msg}\nProntuário:{txt}\n{final_remarks}"

    response = ollama.generate(prompt=prompt, model=model_name)

    print(f"Prontuário {i}:\n{response['response']}")
```
